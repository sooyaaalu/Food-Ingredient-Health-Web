```python
%%time

# 导入相关的包
from paddleocr import PaddleOCRVL
from pathlib import Path
from modelscope import AutoModelForCausalLM, AutoTokenizer
from IPython.display import clear_output

# 全局变量
pipeline = PaddleOCRVL(pipeline_version="v1.6")
model_name = "Qwen/Qwen3-0.6B"
# load the tokenizer and the model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)


def img_2_txt(model_pipeline, img_path):
    output = model_pipeline.predict(img_path)
    for res in output:
        res.save_to_markdown("output.md")
    # print('识别成功，已经文字保存至output文件夹下')


def read_md_join(filename: str) -> str:
    """
    逐行读取 .md 文件，将空白行替换为 '、'，非空白行原样保留，
    最终拼接成一个字符串返回。

    :param filename: .md 文件路径
    :return: 融合后的字符串
    """
    parts = []
    with open(filename, 'r', encoding='utf-8') as f:
        for line in f:
            # 去掉行尾换行符（兼容 \n 与 \r\n）
            content = line.rstrip('\r\n')
            if content.strip() == '':      # 空白行
                parts.append('、')
            else:                          # 非空白行
                parts.append(content)
    return ''.join(parts)


def find_information(prompt = "Give me a short introduction to large language model."):
    # prepare the model input
    
    messages = [
        {"role": "user", "content": prompt}
    ]
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
        enable_thinking=False # Switches between thinking and non-thinking modes. Default is True.
    )
    model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

    # conduct text completion
    generated_ids = model.generate(
        **model_inputs,
        max_new_tokens=32768
    )
    output_ids = generated_ids[0][len(model_inputs.input_ids[0]):].tolist() 

    # parsing thinking content
    try:
        # rindex finding 151668 (</think>)
        index = len(output_ids) - output_ids[::-1].index(151668)
    except ValueError:
        index = 0

    thinking_content = tokenizer.decode(output_ids[:index], skip_special_tokens=True).strip("\n")
    content = tokenizer.decode(output_ids[index:], skip_special_tokens=True).strip("\n")

    # print("thinking content:", thinking_content)
    # print("content:", content)
    return content

def img_2_peiliao(img_path = "shiyan1.jpeg"):
    img_2_txt(pipeline, img_path)
    result = read_md_join('output.md')

    prompt = """
    # Role
    你是一个专业的食品合规与数据分析专家，擅长从非结构化的商品描述文本中精准提取食品配料表信息。

    # Task
    请阅读用户提供的食品描述文本，提取其中的所有配料（包括原辅料和食品添加剂），并严格按照指定格式输出。

    # Rules
    1. **精准提取**：仅提取属于“配料”范畴的成分。坚决忽略净含量、过敏原提示、营养成分、食用方法、保质期、执行标准等无关信息。
    2. **格式严格**：
    - 必须以“配料表：”开头。
    - 各个配料元素之间**只能使用中文顿号（、）** 进行分隔，绝对不能使用逗号、空格、换行符或其他标点符号。
    - 结尾不要加句号。
    3. **复合配料处理**：如果遇到带有括号的复合配料（如“起酥油（氢化植物油、TBHQ）”），请保留其完整性作为单一元素提取，不要强行拆分括号内的内容。
    4. **缺失处理**：如果文本中未包含任何配料信息，请直接输出：“配料表：无”。
    5. **静默输出**：只输出最终格式化的结果，**禁止**包含任何分析过程、解释说明、前言后语或多余的换行。

    # Examples
    ### Example 1
    输入：这款全麦面包非常健康，净含量400g。配料：全麦粉、小麦粉、水、白砂糖、酵母、食用盐、脱氢乙酸钠。过敏原信息：含有麸质。保质期常温15天。
    输出：配料表：全麦粉、小麦粉、水、白砂糖、酵母、食用盐、脱氢乙酸钠

    ### Example 2
    输入：纯天然果汁，主要成分是水、浓缩苹果汁，加了点果葡糖浆。另外加了柠檬酸调节酸度。
    输出：配料表：水、浓缩苹果汁、果葡糖浆、柠檬酸

    ### Example 3
    输入：网红薯片，口感酥脆。执行标准QB/T 2683。营养成分表：能量2000kJ，脂肪30g。致敏物质：大豆制品。
    输出：配料表：无

    ### Example 4
    输入：配料包括优质小麦粉、精炼植物油、复合调味料（味精、食用盐、香辛料）、抗氧化剂（TBHQ）。
    输出：配料表：优质小麦粉、精炼植物油、复合调味料（味精、食用盐、香辛料）、抗氧化剂（TBHQ）

    # Input
    """
    prompt = prompt + result
    clear_output()
    return find_information(prompt = prompt)


img_2_peiliao('ten.png')
```
